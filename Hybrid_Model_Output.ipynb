```python
# Install dependencies
!pip install openai-whisper moviepy transformers scikit-learn matplotlib

import os
import torch
import whisper
import time
import numpy as np
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
from moviepy.editor import VideoFileClip
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt

# Load Whisper
whisper_model = whisper.load_model("small")

# Load models
BART_MODEL = "facebook/bart-large-cnn"
LAMINI_MODEL = "MBZUAI/LaMini-T5-738M"
device = "cuda" if torch.cuda.is_available() else "cpu"

bart_tokenizer = AutoTokenizer.from_pretrained(BART_MODEL)
bart_model = AutoModelForSeq2SeqLM.from_pretrained(BART_MODEL).to(device)

lamini_tokenizer = AutoTokenizer.from_pretrained(LAMINI_MODEL)
lamini_model = AutoModelForSeq2SeqLM.from_pretrained(LAMINI_MODEL).to(device)

MODEL_SIZES_MB = {
    "facebook/bart-large-cnn": 1626,
    "MBZUAI/LaMini-T5-738M": 1475,
}

# Functions
def extract_and_transcribe(video_path):
    audio_path = video_path.replace(".mp4", ".wav")
    video = VideoFileClip(video_path)
    video.audio.write_audiofile(audio_path, codec='pcm_s16le')
    result = whisper_model.transcribe(audio_path)
    os.remove(audio_path)
    return result["text"]

def summarize_bart(text):
    inputs = bart_tokenizer("summarize: " + text, return_tensors="pt", max_length=512, truncation=True).to(device)
    summary_ids = bart_model.generate(**inputs, max_length=250)
    return bart_tokenizer.decode(summary_ids[0], skip_special_tokens=True)

def refine_summary(text):
    inputs = lamini_tokenizer("refine: " + text, return_tensors="pt", max_length=512, truncation=True).to(device)
    summary_ids = lamini_model.generate(**inputs, max_length=200)
    return lamini_tokenizer.decode(summary_ids[0], skip_special_tokens=True)

def compute_metrics(transcript, summary, summarize_time, refine_time):
    original_tokens = bart_tokenizer(transcript, return_tensors="pt", max_length=512, truncation=True).to(device)
    summary_tokens = bart_tokenizer(summary, return_tensors="pt", max_length=512, truncation=True).to(device)

    original_embeds = bart_model.base_model.get_input_embeddings()(original_tokens.input_ids).detach().cpu().numpy()
    summary_embeds = bart_model.base_model.get_input_embeddings()(summary_tokens.input_ids).detach().cpu().numpy()

    original_avg = original_embeds.mean(axis=1)
    summary_avg = summary_embeds.mean(axis=1)
    output_quality = cosine_similarity(original_avg, summary_avg).mean()

    model_size_mb = MODEL_SIZES_MB[BART_MODEL] + MODEL_SIZES_MB[LAMINI_MODEL]
    comp_efficiency = summarize_time + refine_time

    return model_size_mb, output_quality, comp_efficiency

def normalize_to_percent(value, max_value):
    if max_value == 0:
        return 0
    return (value / max_value) * 100

# Video paths
video_folder = "/content/drive/MyDrive/video"
video_paths = [os.path.join(video_folder, f) for f in os.listdir(video_folder) if f.endswith(".mp4")]

# First pass: collect max values for normalization
temp_model_sizes, temp_efficiencies = [], []

for video_path in video_paths:
    transcript = extract_and_transcribe(video_path)

    start_sum = time.time()
    summary_bart = summarize_bart(transcript)
    summarize_time = time.time() - start_sum

    start_ref = time.time()
    refined_summary = refine_summary(summary_bart)
    refine_time = time.time() - start_ref

    model_size, _, comp_efficiency = compute_metrics(transcript, refined_summary, summarize_time, refine_time)
    temp_model_sizes.append(model_size)
    temp_efficiencies.append(comp_efficiency)

max_model_size = max(temp_model_sizes) if temp_model_sizes else 1
max_efficiency = max(temp_efficiencies) if temp_efficiencies else 1

# Second pass: compute and normalize metrics
model_sizes = []
output_qualities = []
computational_efficiencies = []

for idx, video_path in enumerate(video_paths, 1):
    print(f"\nProcessing Video {idx}: {video_path}")
    transcript = extract_and_transcribe(video_path)

    start_sum = time.time()
    summary_bart = summarize_bart(transcript)
    summarize_time = time.time() - start_sum

    start_ref = time.time()
    refined_summary = refine_summary(summary_bart)
    refine_time = time.time() - start_ref

    model_size, output_quality, comp_efficiency = compute_metrics(transcript, refined_summary, summarize_time, refine_time)

    model_size_percent = normalize_to_percent(model_size, max_model_size)
    comp_eff_percent = 100 - normalize_to_percent(comp_efficiency, max_efficiency)
    output_quality_percent = min(output_quality * 100, 100)

    print(f"Refined Summary:\n{refined_summary}")
    print(f"Model Size (%): {model_size_percent:.2f}")
    print(f"Output Quality (%): {output_quality_percent:.2f}")
    print(f"Computational Efficiency (%): {comp_eff_percent:.2f}")
    print("-" * 40)

    model_sizes.append(model_size_percent)
    output_qualities.append(output_quality_percent)
    computational_efficiencies.append(comp_eff_percent)

# Plot
plt.figure(figsize=(12, 6))
x = list(range(1, len(model_sizes) + 1))

plt.plot(x, model_sizes, label="Model Size (%)", marker="o")
plt.plot(x, output_qualities, label="Output Quality (%)", marker="s")
plt.plot(x, computational_efficiencies, label="Computational Efficiency (%)", marker="^")

plt.xlabel("Video Number")
plt.ylabel("Normalized Metric Value (%)")
plt.title("Summarization Performance Metrics (Normalized)")
plt.legend()
plt.xticks(x)
plt.grid(True)
plt.tight_layout()
plt.show()
```
